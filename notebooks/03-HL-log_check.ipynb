```python
with open("../logs/20260804_1138_extract.log") as f:
    lines = [x[10:] for x in f.readlines()]

types = {
    "slash": [],
    "reference": [],
    "ref_plus_strong": [],
    "ref_plus_weak": [],
    "ref_plus_multiline": [],
    "ref_note": [],
    "missing_query": [],
    "plus_in_yref": [],
    "new_entry_from_plus": [],
    "quarter_inch_plus": []
}

for l in lines:
    if l.startswith("slash"):
        types["slash"].append(l[18:])
    elif l.startswith("reference simple"):
        types["reference"].append(l[18:])
    elif l.startswith("reference with plus, strong"):
        types["ref_plus_strong"].append(l)
    elif l.startswith("reference with plus, weak"):
        types["ref_plus_weak"].append(l)
    elif l.startswith("reference with plus, multiline"):
        types["ref_plus_multiline"].append(l)
    elif l.startswith("reference includes"):
        types["ref_note"].append(l)
    elif l.startswith("Queried df"):
        types["missing_query"].append(l)
    elif l.startswith("Y doc ref"):
        types["plus_in_yref"].append(l)
    elif l.startswith("creating new"):
        types["new_entry_from_plus"].append(l)
    elif l.startswith("Quarter Inch ref with +"):
        types["quarter_inch_plus"].append(l)
```
