```python
with open("../logs/20260602_1227_main.log") as f:
    lines = [x[10:] for x in f.readlines()]

types = {"slash": [], "reference": [], "ref_plus": [], "ref_note": []}
for l in lines:
    if "slash" in l:
        types["slash"].append(l[18:])
    if "reference simple" in l:
        types["reference"].append(l[18:])
    if "reference with" in l:
        types["ref_plus"].append(l)
    if "reference includes" in l:
        types["ref_note"].append(l)
```
